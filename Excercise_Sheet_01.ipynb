### Problem 1 (code)

### Problem 2 (code)

```python
m = 1
x = 5

# printing lines
while m <= x:
    n = 1
    # printing numbers in the lines plus an empty space
    while n <= m:
        print(n, end=" ")
        n = n + 1
    print("")
    m = m + 1
```

### Problem 3 (code)

### Problem 4 (code)

Write code to print the formatted multiplication table for a given number n.

n = 1

1 x 1 = 1
1 x 2 = 2
1 x 3 = 3

```python
n = 1
m = 3
o = 1

while n <= m:
    print(o, "x", n , "=", o * n)
    n = n + 1
```

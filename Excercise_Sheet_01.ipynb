### Problem 7

Write code to print all prime numbers between any given pair of numbers n and m to the console.
Use as example input n = 25 and m = 50.
NOTE:A prime number is a natural number greater than 1 that can only be divided by itself and
1 without remainders. Examples are: 2, 3, 5, 7, 11.


```python
n = 1
m = 10

# sequence = list(range(100))  # Example sequence
# result = sequence[n:m]

for i in range(n, m + 1):
    if i < 2:
        continue
    elif i % 2 == 0 and i != 2:
        continue
    prime = True
    for div in range(2, int(i ** 0.5) + 1):
        if i % div == 0:
            prime = False
            break
    if prime:
        print(i)

print("")

for i in range(n, m + 1):
    # check if i is prime iterating over each number from i down to 1
    j = i - 1
    while j > 1:
        if i % j == 0:
            break
        j = j - 1

    if j == 1:
        print(i)

```

### Problem 8

Write code that reverses a given string number s. For example “stressed“ should result in “desserts“.
Similarly, write code that reverses a given integer n. For example n = 76542 should result in m =
24567.

```python
seq = "stressed"
num = str(76542)


rev = seq[::-1]
revnum = num[::-1]

print(rev)
print(int(revnum))

def invert_integer(n):
    res = 0
    while n > 0:
        m = n % 10
        res = res * 10 + m
        n = n // 10
    return res

print (invert_integer(76542))

# universal invert function that can handle both strings and integers
def invert(n):
    if type(n) == int:
        return invert_integer(n)
    else:
        return n[::-1]

print(invert("stressed"))
print(invert(76542))

```
